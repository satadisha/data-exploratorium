# DS 3100 Activity Solution: Exploring the Tennessee Education Data

This notebook provides a solution to the data-frame activity using **R** and the provided `tenn2018.csv` file.

> **Note:** Place `tenn2018.csv` in a data folder that is in the same directory as this notebook before running it.

## 1. Read the data and print the first few observations

We use `read.csv()` to read the CSV file into an R **data frame** called `ten`. The `head()` function displays the first six observations by default.

```r
ten <- read.csv("data/tenn2018.csv")
head(ten)
```

The first few rows show the structure of the dataset: each row is an observation, while each column is a variable.

## 2. Determine the number of rows and columns

`nrow()` returns the number of observations (rows), and `ncol()` returns the number of variables (columns).

```r
nrow(ten)
ncol(ten)
```

**Answer:** The dataset contains **121,062 rows** and **18 columns**.

## 3. Print the column names

`names()` returns the names of the variables in the data frame. The length of this vector is the number of columns.

```r
cat("Column names: ")
cat(names(ten), sep = ", " , "\n")
cat("Number of columns:", length(names(ten)), "\n")
```

**Answer:** There are **18 variables** in the dataset. Looking into the data, there appears to be 4 columns holding categorical values.

## 4. Access the `overall_subject` column

The `$` operator is a convenient way to access a named column of a data frame. The expression below returns the complete `overall_subject` column.

```r
ten$overall_subject
```

## 5. Access `district_name` for the first observation

A data frame can be indexed using `[row, column]`. R uses **1-based indexing**, so the first observation is row `1`.

```r
ten[1, "district_name"]
```

**Answer:** `"State of Tennessee"`

## 6. Access `percent_below` for the first observation

```r
ten[1, "percent_below"]
```

**Answer:** `19.5`

## 7. How many columns contain missing values?

First, `is.na(ten)` creates a logical table indicating where values are missing.

`colSums(is.na(ten))` counts the missing values in each column. We then check which columns have **at least one** missing value and count them.

```r
missing_by_column <- colSums(is.na(ten))

cat("Number of columns with missing values:", sum(missing_by_column > 0), "\n")
```

**Answer:** **15 columns** contain at least one missing value.

The other **3 columns**—`district_number`, `school_number`, and `year`—have no missing values.
