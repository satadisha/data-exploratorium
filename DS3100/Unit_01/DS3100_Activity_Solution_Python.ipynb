# DS 3100 Activity Solution: Exploring the Tennessee Education Data

This notebook provides a solution to the data-frame activity using **Python/pandas** and the provided `tenn2018.csv` file.

> **Note:** Place `tenn2018.csv` in a data folder that is in the same directory as this notebook before running it.

## 1. Read the data and print the first few observations

We use `pd.read_csv()` to read the CSV file into a pandas **DataFrame** called `ten`. The `head()` method displays the first five observations by default.

```python
import pandas as pd

ten = pd.read_csv("data/tenn2018.csv")
ten.head()
```

The first few rows show the structure of the dataset: each row is an observation, while each column is a variable.

## 2. Determine the number of rows and columns

The DataFrame attribute `.shape` returns `(number_of_rows, number_of_columns)`.

```python
ten.shape
```

**Answer:** The dataset contains **121,062 rows** and **18 columns**.

## 3. Print the column names

`ten.columns` returns the names of all variables in the DataFrame. `len(ten.columns)` gives the number of variables.

```python
print("Column names:", ten.columns.tolist())
print("Number of columns:", len(ten.columns))
```

**Answer:** There are **18 variables** in the dataset. Looking into the data, there appears to be 4 columns holding categorical values.

## 4. Access the `overall_subject` column

In pandas, square brackets can be used to select a column by name. `ten["overall_subject"]` returns the complete column as a pandas Series.

```python
ten["overall_subject"]
```

## 5. Access `district_name` for the first observation

A pandas DataFrame can be indexed by row and column. Using `.loc` lets us specify the row and column labels explicitly. With the default integer index, the first observation has index `0`.

```python
ten.loc[0, "district_name"]
```

**Answer:** `"State of Tennessee"`

## 6. Access `percent_below` for the first observation

```python
ten.loc[0, "percent_below"]
```

**Answer:** `19.5`

## 7. How many columns contain missing values?

`ten.isna()` creates a Boolean DataFrame showing where values are missing.

`.any()` checks whether each column contains at least one missing value, and `.sum()` counts how many columns satisfy that condition.

```python
missing_by_column = ten.isna().any(axis=0).sum()
print("Number of columns with missing values:", missing_by_column)
```

**Answer:** **15 columns** contain at least one missing value.

The other **3 columns**—`district_number`, `school_number`, and `year`—have no missing values.
