# What is this course about?

An introduction to the <span style="color: #1565C0;">**detailed foundations of data science**</span> and to the practical methods and tools used in data analysis.

We will learn how to move from a real-world question to a systematic, reproducible analysis:

- acquire and organize data
- clean and wrangle data
- explore and visualize data
- build and evaluate models
- interpret results
- communicate meaningful insights

The course also introduces responsible data science, discussing topics like <span style="color: #1f806c;">**ethics, privacy, and reproducibility**</span>.

# 💻 What about coding?

We want to implement things but this course is <span style="color:#1f806c;">**not about choosing the “right” programming language**</span>.

We will use both **Python and R** to implement the same data science ideas.


- The statistical and analytical concepts remain the same.
- The workflow remains the same.

- The syntax and libraries differ.
- You may use either language for Labs and Analysis Reports.


## 🐍 Why Python?

Python is a general-purpose programming language with a large ecosystem for data science.

In this course, Python gives us a practical environment for:


- working with tabular data using <span style="color:#1f806c;">**pandas**</span>
- numerical computing with <span style="color:#1f806c;">**NumPy**</span>
- visualization with <span style="color:#1f806c;">**matplotlib / seaborn**</span>
- statistical modeling and machine learning with <span style="color:#1f806c;">**statsmodels / scikit-learn**</span>
- building reproducible computational workflows in notebooks


## 📊 Why R?

R was designed for statistical computing and data analysis and has a particularly strong ecosystem for exploratory analysis, visualization, and statistical modeling.

In this course, R gives us a complementary way to:


- manipulate tidy data with <span style="color:#1f806c;">**dplyr / tidyr**</span>
- create visualizations with <span style="color:#1f806c;">**ggplot2**</span>
- build statistical models with base R and <span style="color:#1f806c;">**tidymodels**</span>
- perform reproducible statistical analyses


## R and Python: the same idea

For example, suppose we want to read a CSV file.

### Python

```python
import pandas as pd

df = pd.read_csv("data.csv")
```

### R

```r
library(readr)

df <- read_csv("data.csv")
```

The syntax differs; the **data science task is the same**.


## R and Python: Comparative Workflow

| Data science task | Python | R |
|---|---|---|
| Read data | `pandas.read_csv()` | `readr::read_csv()` |
| Filter rows | `query()` / boolean indexing | `dplyr::filter()` |
| Create variables | assignment / `assign()` | `dplyr::mutate()` |
| Group and summarize | `groupby()` | `group_by()` + `summarize()` |
| Visualize | `matplotlib` / `seaborn` | `ggplot2` |
| Linear regression | `statsmodels` / `sklearn` | `lm()` / `tidymodels` |


# What matters more than syntax

Data science is more than programming. It is the process of turning **data into evidence, evidence into insight, and insight into useful decisions**.

The programming language is a **means to those ends**.

In DS 3100, we will learn that process using two complementary programming ecosystems: **Python and R**.

## Overall choice

- <span style="color:#1565C0;">**Choose Python**</span> for versatility, machine learning, automation, deploying models into applications and integration with other software.
- <span style="color:#1565C0;">**Choose R**</span> for specialized statistical analysis, academic research, and high-quality data visualization.

# How programming fits into the course

We will use <span style="color:#1f806c;">**Visual Studio Code**</span> as the primary development environment for both Python and R.

We will also use <span style="color:#1f806c;">**Jupyter notebooks**</span> for computational work.

Students may complete labs and Analysis Reports in either Python or R; instruction and grading will support both languages.
